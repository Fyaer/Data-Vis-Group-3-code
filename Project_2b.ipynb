```python
#Importing Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#switching from matplotlib's default graphing style to that of seaborn's. Seaborn displays graph lines, matplotlib doesn't
sns.set(color_codes = True)

#removing warnings from the output
import warnings
warnings.filterwarnings("ignore")
```

```python
#importing Dataset
fish = pd.read_csv('Fish.csv')
```

```python
#peeking at fish data
fish.head()
fish.columns
```

```python
#looking at records for each fish species
fish["Species"].value_counts()
```

```python
fish.plot(kind='scatter', x = 'Weight', y = 'Length1')
plt.title('Weight vs Length1')
```

```python
rp=sns.jointplot(x = 'Length1', y = 'Width', data = fish, height =5)
rp.fig.subplots_adjust(top=0.9) # adjust the Figure in rp
rp.fig.suptitle('Jointplot showing  Fish length vs width')
plt.show()
```

```python
rp=sns.FacetGrid(fish, hue='Species', palette='husl', height=5).map(plt.scatter, 'Length1', 'Width').add_legend()
rp.fig.subplots_adjust(top=0.9) # adjust the Figure in rp
rp.fig.suptitle('Jointplot showing  Fish length vs  width')
plt.show()
```

```python
sns.boxplot(x='Species', y='Width', palette='husl', data=fish).set(title='Boxplot showing fish species vs width')
plt.show()
```

```python
sns.boxplot(x='Species', y ='Length1', palette='husl', data = fish).set(title='Boxplot showing Fish Species vs Length')
plt.show()
```

```python
sns.violinplot(x='Species', y='Width', palette='husl', data=fish).set(title='Violinplot showing Fish Species vs  Width')
plt.show()
```

```python
# Distribution Plot (a.k.a. Histogram)
sns.distplot(fish['Length1']).set(title='Histogram showing Length1 of Fish data')
plt.show()
```

```python
sns.FacetGrid(fish, hue='Species', palette='husl', height=6).map(sns.kdeplot, 'Length1').add_legend().set(title='Kdeplot showing Fish Length1 vs density')
plt.show()
```

```python
numeric_fish = fish.select_dtypes(include=['float', 'int'])

corr = numeric_fish.corr()

# Heatmap
sns.heatmap(corr).set(title='Correlation plot of Fish Dataset')
plt.show()
```

```python
colors = ['#78C850', '#F08030','#6890F0']
# Count Plot (a.k.a. Bar Plot)
sns.countplot(x='Species', data=fish, palette=colors)

# Rotate x-labels
plt.xticks(rotation=-45)
```

```python
# Factor Plot
rp = sns.catplot(x='Width',
                   y='Length1',
                   data=fish,
                   hue='Species',  # Color by species
                   col='Species',  # Separate by species
                   kind='swarm')  # Swarmplot

# Rotate x-axis labels
rp.fig.subplots_adjust(top=0.9) # adjust the Figure in rp

rp.set_xticklabels(rotation=90)
rp.fig.suptitle('Factorplot showing Fish Width vs Fish Length')
plt.show()
```

```python
from pandas.plotting import parallel_coordinates
parallel_coordinates(fish, 'Species')
plt.title('Parallel Coordinates')
plt.show()
```
